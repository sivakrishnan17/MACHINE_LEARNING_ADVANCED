```python
import pandas as pd
import numpy as np
```

```python
load=pd.read_csv('pre_placement.csv')
```

```python
from sklearn.feature_selection import SelectKBest,chi2,RFE
```

```python
from sklearn.model_selection import train_test_split
```

```python
load.head()
```

```python
load=pd.get_dummies(load,drop_first=True).astype(int)
```

```python
load.tail()
```

```python
x=load.drop('salary',axis=1)
y=load['salary']
```

```python
kbest=SelectKBest(score_func=chi2,k=7)
fits=kbest.fit(x,y)
x=fits.transform(x)
```

```python
kbest.scores_
```

```python
kbest.get_support()
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
dt=LinearRegression()
dt.fit(xtrain,ytrain)
```

```python
ypred=dt.predict(xtest)
```

```python
from sklearn.metrics import r2_score
r2=r2_score(ytest,ypred)
```

```python
r2
```

```python
model_score=[]
model_score.append(r2)
```

```python
df=pd.DataFrame(index=['r2'],columns=['multi linear regression'])
```
